# Split Data by Size Bucket

```python
from pathlib import Path
import sys
import pandas as pd
import numpy as np

# Face-Signal root
ROOT = Path().resolve().parents[1]
sys.path.append(str(ROOT))

%load_ext autoreload
%autoreload 2
```

```python
DATA_DIR = ROOT / "data" / "metadata"
```

```python
import pandas as pd

df = pd.read_csv(DATA_DIR / "face_image_sizes_annotated.csv")

cols = ['image_path', 'size_bucket', 'Target', 'Type', 'Model', 'Skin Tone', 'Skin Tone Group',
       'Predicted Gender', 'Predicted Age', 'Intersection']

df = df[cols]
df.head()
```

```python
bucket_dfs = {}

for bucket, group in df.groupby('size_bucket'):
    safe_name = bucket.replace('(', '').replace(']', '').replace(', ', '_')
    path = DATA_DIR / f'bucket_{safe_name}.csv'
    group.to_csv(path, index=False)
    bucket_dfs[bucket] = group
    print(f'{bucket:>15}  |  {len(group):>6}  →  {path.name}')
```
