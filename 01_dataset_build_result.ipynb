# Dataset Construction Results

### 01. Create face, skin dataset list from full dataset list

```python
import pandas as pd
from pathlib import Path

full_df = pd.read_csv('outputs/full_dataset.csv')
```

1-1) Face Dataset:

```python
face_df = full_df.copy()

face_df["image_path"] = face_df["image_path"].str.replace(
    "AI_Face_imagesV2", "AI_Face_imagesV2_face", regex=False
)

face_df["Image Path"] = face_df["Image Path"].str.replace(
    "AI_Face_imagesV2", "AI_Face_imagesV2_face", regex=False
)

face_df[["Image Path", "image_path"]].head()
```

1-2) Skin Dataset:

```python
skin_df = full_df.copy()

skin_df["image_path"] = skin_df["image_path"].str.replace(
    "AI_Face_imagesV2", "AI_Face_imagesV2_skin", regex=False
)

skin_df["Image Path"] = skin_df["Image Path"].str.replace(
    "AI_Face_imagesV2", "AI_Face_imagesV2_skin", regex=False
)

skin_df[["Image Path", "image_path"]].head()
```

```python
# face_df.to_csv("outputs/face_dataset.csv", index=False)
# skin_df.to_csv("outputs/skin_dataset.csv", index=False)
```

### 02. Processed image path check

```python
face_df["exists"] = face_df["image_path"].apply(lambda p: Path(p).exists())

print(face_df.head())
```

```python
face_df["exists"].value_counts(normalize=True)
```

```python
face_df_valid = face_df[face_df["exists"] == True]
face_df_valid.to_csv("outputs/face_dataset_valid.csv", index=False)
```

2-2) Skin Dataset:

```python
skin_df["exists"] = skin_df["image_path"].apply(lambda p: Path(p).exists())

print(skin_df.head())
```

```python
skin_df["exists"].value_counts(normalize=True)
```

```python
skin_df_valid = skin_df[skin_df["exists"] == True]
skin_df_valid.to_csv("outputs/skin_dataset_valid.csv", index=False)
```
